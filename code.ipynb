```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

col_name  = ['gender','Residence_type', 'ever_married']

for col in col_name:
    dataset[col] = le.fit_transform(dataset[col])
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

col1_name = ['work_type','smoking_status']

ct = ColumnTransformer(transformers=[('encoder',OneHotEncoder(drop='first'), col1_name)],remainder='passthrough')
```

```python
X = ct.fit_transform(dataset.drop(columns=['id','stroke']))
y = dataset['stroke'].values
```

```python
X
```

```python
y
```

```python
from  sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier,GradientBoostingClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.naive_bayes import GaussianNB

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

from sklearn.metrics import confusion_matrix,accuracy_score
```

```python
models = {
    "Logistic Regression": LogisticRegression(),
    "K-Nearest Neighbors": KNeighborsClassifier(),
    "Support Vector Machine": SVC(),
    "Decision Tree": DecisionTreeClassifier(),
    "Random Forest": RandomForestClassifier(),
    "Gradient Boosting": GradientBoostingClassifier(),
    "Naive Bayes": GaussianNB(),
    "XGBoost": XGBClassifier(use_label_encoder=False, eval_metric='logloss'),
    "LightGBM": LGBMClassifier()
    }
```

```python
results = {}

for name,model in models.items():
    model.fit(X_train,y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test,y_pred)
    results[name] = acc
    print(f"{name} Accuracy:{acc : 4f}")
```

```python
import pandas as pd

results_df = pd.DataFrame(list(results.items()), columns=["Model", "Accuracy"])
results_df = results_df.sort_values(by="Accuracy", ascending=False)
print(results_df)
```

```python
ann = Sequential()

ann.add(Dense(units=8,activation='relu',input_dim=X_train.shape[1]))

ann.add(Dense(units=8,activation='relu'))

ann.add(Dense(units=1,activation='sigmoid'))

ann.compile(optimizer='adam',loss= 'binary_crossentropy',metrics=['accuracy'])

ann.fit(X_train,y_train,batch_size=32,epochs=50,verbose=0)

y_pred_ann = (ann.predict(X_test)>0.5).astype('int32')
acc_ann = accuracy_score(y_test,y_pred_ann)
results["ANN"] = acc_ann

```

```python
results_df = pd.DataFrame(list(results.items()), columns=["Model", "Accuracy"])
results_df = results_df.sort_values(by="Accuracy", ascending=False)
print(results_df)
```
